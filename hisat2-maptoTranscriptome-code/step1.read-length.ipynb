```python
# get read length from hisat2 output sam-format data
def readlength(inputFile,outputFile):
    # open file
    outfile = open(outputFile,'w')
    len_dict = {}
    with open(inputFile,'r') as input:
        for line in input:
            if line.startswith('@'):
                continue 
            fileds = line.split()
            if fileds[2] == '*':
                continue
            length = len(fileds[9])
            len_dict.setdefault(length,0)
            len_dict[length] += 1
            
    # sort dict
    len_dict_sorted = list(len_dict.items())
    len_dict_sorted.sort()
    
    # output
    for elem in len_dict_sorted:
        outfile.write('\t'.join([str(elem[0]),str(elem[1])]) + '\n')
    outfile.close()
```
